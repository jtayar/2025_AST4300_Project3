## Task 4: What does the profile look like?

You'll see lots of plots of stellar evolution tracks, but having the models also allows you to see what's really going on inside. To do that, you need to load in the profile files

We are going to pick a profile close to the solar age, 4.5 Gyrs for this one solar mass solar metallicity model. 

```python
profilelist1=mr.MesaLogDir('MESA-Web_Job_09152553902', history_file='trimmed_history.data')
```

```python
def is_old_and_bright(age, log_lum):
     return age > 3e9 and log_lum > 0
m_nums = profilelist1.select_models(is_old_and_bright, 'star_age', 'log_L')
print(m_nums)
```

```python
profile1_solar=profilelist1.profile_data(300) #change this number for your model?
```
